#  Iterators 

##  Definition

An **iterator** is an object in Python that allows you to traverse (loop through) all elements of a collection **one by one**.

- In simple words:
It is an object that **remembers its position and returns the next item when requested**.

---

#  Iterator vs Iterable

##  Iterable

* An object that can be looped over
* Examples: list, tuple, string

```python
nums = [1, 2, 3]
```

---

##  Iterator

* Object that actually performs iteration
* Has methods:

  * `__iter__()`
  * `__next__()`

---

#  Creating an Iterator

```python
nums = [1, 2, 3]

it = iter(nums)

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
```

---

#  StopIteration

```python
print(next(it))  # Error
```

- Raises:

```
StopIteration
```

---

#  Using Iterator in Loop

```python
nums = [1, 2, 3]

for num in nums:
    print(num)
```

- Internally, `for` loop uses iterator

---

#  How `for` Loop Works Internally

```python
it = iter(nums)

while True:
    try:
        item = next(it)
        print(item)
    except StopIteration:
        break
```

---

#  Creating Custom Iterator

```python
class Count:
    def __init__(self, max):
        self.max = max
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.max:
            self.current += 1
            return self.current
        else:
            raise StopIteration
```

---

##  Example Usage

```python
counter = Count(3)

for num in counter:
    print(num)
```

### Output:

```
1
2
3
```

---

#  Iterator Protocol

An object is an iterator if it implements:

* `__iter__()` → returns iterator object
* `__next__()` → returns next value

---

#  Advantages

1. Saves memory (lazy evaluation)
2. Efficient for large data
3. Works one element at a time

---

#  Disadvantages

1. Cannot go backward
2. Cannot reuse once exhausted
3. Slightly complex to implement

---

#  Important Notes 

* Iterable ≠ Iterator
* Use `iter()` to create iterator
* Use `next()` to get next element
* Raises `StopIteration` when finished
* `for` loop automatically uses iterator

---

#  Summary

* Iterator allows sequential access
* Uses `__iter__()` and `__next__()`
* Works element by element
* Used internally by loops
