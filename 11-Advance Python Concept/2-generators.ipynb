#  Generators

##  Definition

A **generator** is a special type of iterator that generates values **one at a time** using the `yield` keyword instead of returning all values at once.

- In simple words:
It **produces values lazily (on demand)** rather than storing them in memory.

---

#  Why Use Generators?

* Saves memory
* Efficient for large data
* Generates values only when needed
* Simpler than creating custom iterators

---

#  Generator Function

- A function becomes a generator when it uses `yield`

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

---

##  Example Usage

```python
gen = my_generator()

print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
```

---

#  Generator vs Function

| Feature   | Function          | Generator            |
| --------- | ----------------- | -------------------- |
| Keyword   | `return`          | `yield`              |
| Execution | Runs completely   | Pauses and resumes   |
| Memory    | Stores all values | Generates one by one |

---

#  Generator with Loop

```python
def count(n):
    for i in range(n):
        yield i
```

---

##  Using with `for` Loop

```python
for num in count(5):
    print(num)
```

---

#  Generator Expression

- Similar to list comprehension but uses `()`

```python
gen = (x**2 for x in range(5))
```

---

##  Example

```python
for value in gen:
    print(value)
```

---

#  How Generator Works

* When `yield` is reached → function pauses
* State is saved
* Next call resumes from last point

---

#  StopIteration

```python
gen = count(2)

print(next(gen))
print(next(gen))
print(next(gen))  # Error
```

- Raises:

```
StopIteration
```

---

#  Advantages

1. Memory efficient
2. Faster for large data
3. Easy to implement
4. Supports lazy evaluation

---

#  Disadvantages

1. Cannot reuse once exhausted
2. Harder to debug
3. No random access

---

#  Important Notes 

* Uses `yield` keyword
* Returns generator object
* Works like iterator
* Saves memory
* Used in large data processing

---

#  Summary

* Generator produces values one by one
* Uses `yield` instead of `return`
* Memory efficient and fast
* Useful for large datasets


````python
def square(n):
    for i in range(3):
        return i**2
````

````python
square(5)
````

````python
def square(n):
    for i in range(n):
        yield i**2
````

````python
square(5)
````

````python
for i in square(5):
    print(i)
````

````python
def my_generator():
    yield 1
    yield 2
    yield 3
````

````python
gen = my_generator()
print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
````

````python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
````

````python
file_path = 'large_file.txt'
for line in read_large_file(file_path):
    print(line.strip())
````
