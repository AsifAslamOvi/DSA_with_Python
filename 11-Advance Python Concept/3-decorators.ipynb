#  Decorators 

##  Definition

A **decorator** is a function that **modifies or extends the behavior of another function** without changing its actual code.

- In simple words:
It **wraps a function and adds extra functionality**.

---

#  Why Use Decorators?

* Code reuse
* Cleaner and more readable code
* Add functionality (logging, authentication, timing, etc.)
* Avoid modifying original function

---

#  Basic Concept

```python id="n8u2x1"
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper
```

---

##  Applying Decorator

```python id="q4m7z3"
@decorator
def say_hello():
    print("Hello")

say_hello()
```

### Output:

```id="k2v9s8"
Before function
Hello
After function
```

---

#  Without `@` Syntax

```python id="t6y3p1"
def say_hello():
    print("Hello")

say_hello = decorator(say_hello)
say_hello()
```

---

#  Decorator with Arguments

```python id="x7k1d9"
def decorator(func):
    def wrapper(name):
        print("Before")
        func(name)
        print("After")
    return wrapper

@decorator
def greet(name):
    print("Hello", name)

greet("Asif")
```

---

#  Using `*args` and `**kwargs`

- Makes decorator flexible

```python id="p3v6r2"
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Before")
        result = func(*args, **kwargs)
        print("After")
        return result
    return wrapper
```

---

#  Returning Values

```python id="c9m4s7"
@decorator
def add(a, b):
    return a + b

print(add(2, 3))
```

---

#  Multiple Decorators

```python id="u2z8k5"
def deco1(func):
    def wrapper():
        print("Deco1")
        func()
    return wrapper

def deco2(func):
    def wrapper():
        print("Deco2")
        func()
    return wrapper

@deco1
@deco2
def say():
    print("Hello")
```

---

#  Built-in Decorators

##  `@staticmethod`

```python id="r1j7q8"
class A:
    @staticmethod
    def show():
        print("Static method")
```

---

##  `@classmethod`

```python id="m8p2x6"
class A:
    @classmethod
    def show(cls):
        print("Class method")
```

---

##  `@property`

```python id="z5n9v3"
class A:
    def __init__(self):
        self._x = 10

    @property
    def x(self):
        return self._x
```

---

#  Advantages

1. Code reuse
2. Cleaner syntax
3. Separation of concerns
4. Easy to extend functionality

---

#  Disadvantages

1. Can be confusing for beginners
2. Debugging becomes harder
3. Overuse reduces readability

---

#  Important Notes 

* Decorators wrap functions
* Use `@decorator_name` syntax
* Use `*args, **kwargs` for flexibility
* Can modify input/output behavior
* Common in real-world applications

---

# Summary

* Decorators modify function behavior
* Implemented using wrapper functions
* Improve code reuse and readability
* Widely used in Python frameworks

````python
'''
Closures -> A closure is a function object that has access to variables in its enclosing scope, even after the outer 
function has finished executing. Closures are created when a nested function references a value in its enclosing 
scope.
'''

def main_welcome():
    msg = "Welcome"

    def Sub_welcome_method():
        print("Welcome to the world of Python programming!")
        print(msg)
        print("Please learn this concept properly")
    
    return Sub_welcome_method

welcome_message = main_welcome()
welcome_message()
````

````python

def main_welcome(func):

    def Sub_welcome_method():
        print("Welcome to the world of Python programming!")
        func("Welcome everyone")
        print("Please learn this concept properly")
    
    return Sub_welcome_method

main_welcome(print)()
````

````python

def main_welcome(func, lst):

    def Sub_welcome_method():
        print("Welcome to the world of Python programming!")
        print(func(lst))
        print("Please learn this concept properly")
    
    return Sub_welcome_method

main_welcome(len, [1, 2, 3, 4, 5])()
````

````python
# Decorator
def decorator_function(func):
    def wrapper_function():
        print("This is a decorator function.")
        func()
        print("This is the end of the decorator function.")
    return wrapper_function

````

````python
def course_introduction():
    print("This is a course on advanced Python concepts.")

course_introduction()
````

````python
decorator_function(course_introduction)()
````

````python
@decorator_function
def course_introduction():
    print("This is a course on advanced Python concepts.")

course_introduction()
````

````python
def my_decorator(func):
    def wrapper():
        print("Before the function call.")
        func()
        print("After the function call.")
    return wrapper

````

````python
@my_decorator
def say_hello():
    print("Hello!")

say_hello()
````

````python
## Decorators with arguments
def repeat(num_times):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                func(*args, **kwargs)
        return wrapper
    return decorator
````

````python
@repeat(3)
def greet(name):
    print(f"Hello, {name}!")

greet("Ovi")
````
