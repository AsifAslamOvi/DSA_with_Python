#  PMI (Principle of Mathematical Induction)

##  Definition

The **Principle of Mathematical Induction (PMI)** is a mathematical proof technique used to prove that a statement is true for **all natural numbers**.

- In simple words:
If a statement is true for the first number and true for the next number whenever it is true for any number, then it is true for all natural numbers.

---

#  When to Use PMI?

* To prove formulas involving natural numbers
* To prove divisibility statements
* To prove inequalities
* To prove recursive relations

---

#  Steps of PMI

To prove a statement ( P(n) ) for all ( n \geq 1 ):

##  Step 1: Base Case

Prove that the statement is true for the first value (usually `n = 1`).

##  Step 2: Inductive Hypothesis

Assume the statement is true for some natural number `k`.

```text
Assume P(k) is true
```

##  Step 3: Inductive Step

Using the assumption, prove that:

```text
P(k + 1) is true
```

##  Step 4: Conclusion

Therefore, the statement is true for all natural numbers.

---

#  General Form

If:

```text
1. P(1) is true
2. P(k) ⇒ P(k+1)
```

Then:

```text
P(n) is true for all n ∈ N
```

---

#  Example 1: Sum of First n Natural Numbers

Prove:

```text
1 + 2 + 3 + ... + n = n(n + 1) / 2
```

##  Base Case (n = 1)

```text
LHS = 1
RHS = 1(1+1)/2 = 1
```

True 

---

##  Inductive Hypothesis

Assume true for `n = k`

```text
1 + 2 + ... + k = k(k+1)/2
```

---

##  Inductive Step

Prove for `k+1`

```text
1 + 2 + ... + k + (k+1)
= k(k+1)/2 + (k+1)
= (k+1)(k/2 + 1)
= (k+1)(k+2)/2
```

Which equals:

```text
(k+1)((k+1)+1)/2
```

True 

---

##  Conclusion

Hence proved by PMI.

---

#  Example 2: Power of 2

Prove:

```text
2^n > n
for all n ≥ 1
```

Use same induction steps.

---

#  Types of Induction

##  1. Simple Induction

Uses `P(k) ⇒ P(k+1)`

##  2. Strong Induction

Assume all previous cases true:

```text
P(1), P(2), ... P(k)
```

Then prove `P(k+1)`

---

#  Advantages

1. Powerful proof method
2. Useful for sequences and formulas
3. Easy logical structure

---

#  Disadvantages

1. Only works for ordered sets like natural numbers
2. Requires correct hypothesis

---

#  Important Notes 

* Always write all 3 steps clearly
* Base case is mandatory
* Hypothesis must be assumed correctly
* Final conclusion must be written

---

#  Summary

* PMI proves statements for all natural numbers
* Uses Base Case + Hypothesis + Inductive Step
* Very important in discrete mathematics
* Common in proofs of sums, divisibility, inequalities
 

````python
def power2(n):
    if(n==1): # base case
        return 2
    
    smallAnswer = power2(n-1) # Assume (2) -> Induction Hypothesis
    ans = 2 * smallAnswer # (3) using (2)

    return ans


print(power2(5))
````

````python
# Sum of n natural numbers using recursion
def sumN(n):
    if(n==1):
        return 1
    
    smallAnswer = sumN(n-1) 
    ans = n + smallAnswer 

    return ans

print(sumN(5))
````

````python
# Number of digits in a number using recursion
def countDigits(n):
    if(n==0):
        return 0
    
    digit = n % 10
    print(f"{digit}\t", end="")
    total_digit = countDigits(n//10)
    return 1 + total_digit

print(f"\nTotal digits: {countDigits(12345)}")
````

````python
# Sum of number of digits in a number using recursion
def sumDigits(n):
    if(n==0):
        return 0
    
    digit = n % 10
    print(f"{digit}\t", end="")
    sum_of_digits = sumDigits(n//10)
    return digit + sum_of_digits

print(f"\nSum of digits: {sumDigits(12345)}")
````

````python
# Fibonacci number using recursion
def fibonacci(n):
    if(n==0):
        return 0
    if(n==1):
        return 1
    
    last = fibonacci(n-1)
    second_last = fibonacci(n-2)

    ans = last + second_last
    return ans

n = 5
print(f"Fibonacci of {n}: {fibonacci(n)}") # 0,1,1,2,3,5
````
