#  Head Recursion vs Tail Recursion 

##  Definition

**Recursion** is a technique where a function calls itself to solve a smaller version of the same problem.

There are two common types:

* **Head Recursion**
* **Tail Recursion**

---

#  1. Head Recursion

##  Definition

A recursive function is called **Head Recursion** when the recursive call happens **before** any processing statement.

- First recursive call happens, then work is done while returning.

---

##  Syntax

```python id="h1r7k2"
def function(n):
    if n == 0:
        return
    function(n-1)
    print(n)
```

---

##  Example

```python id="d3p8m4"
def head_recursion(n):
    if n == 0:
        return
    head_recursion(n-1)
    print(n)

head_recursion(5)
```

### Output

```text id="s8x4t1"
1
2
3
4
5
```

---

##  Working

For `n = 5`

```text id="p9w2c6"
head_recursion(5)
→ head_recursion(4)
→ head_recursion(3)
→ head_recursion(2)
→ head_recursion(1)
→ head_recursion(0)

Then prints:
1 2 3 4 5
```

---

#  2. Tail Recursion

##  Definition

A recursive function is called **Tail Recursion** when the recursive call happens **last**, after processing statements.

- Work is done first, then recursive call happens.

---

##  Syntax

```python id="f6m1z9"
def function(n):
    if n == 0:
        return
    print(n)
    function(n-1)
```

---

##  Example

```python id="u7q3v5"
def tail_recursion(n):
    if n == 0:
        return
    print(n)
    tail_recursion(n-1)

tail_recursion(5)
```

### Output

```text id="r2k9n7"
5
4
3
2
1
```

---

##  Working

For `n = 5`

```text id="c5x8b2"
print(5)
tail_recursion(4)

print(4)
tail_recursion(3)

print(3)
tail_recursion(2)
...
```

---

#  Comparison Table

| Feature                 | Head Recursion       | Tail Recursion        |
| ----------------------- | -------------------- | --------------------- |
| Recursive Call Position | First                | Last                  |
| Processing Happens      | After recursive call | Before recursive call |
| Output Order            | Ascending            | Descending            |
| Example Output          | 1 2 3 4 5            | 5 4 3 2 1             |

---

#  Python Note

- Python does **not optimize tail recursion**.
So tail recursion does **not save memory** like some other languages.

---

#  Advantages

## Head Recursion

1. Useful when result needed after return
2. Natural for reverse operations

## Tail Recursion

1. Cleaner logic in some problems
2. Easier to convert into loops

---

#  Important Notes 

* Head recursion → recursive call first
* Tail recursion → recursive call last
* Python supports both
* No tail-call optimization in Python

---

#  Summary

* **Head Recursion**: call first, work later
* **Tail Recursion**: work first, call later
* Both are recursive techniques used to solve problems recursively


````python
def fact_head(n):
    if n == 0:
        return 1
    
    small_answer = fact_head(n-1)
    final_answer = n * small_answer
    return final_answer

print(fact_head(5))
````

````python
def fact_tail(n):
    if n == 0:
        return 1
    
    return n * fact_tail(n-1)

print(fact_tail(5))
````

````python
def fact_tail(n, accumulator=1):
    if n == 0:
        return accumulator
    
    return fact_tail(n-1, accumulator * n)

'''
accumulator is a variable that stores the intermediate result of the factorial calculation.
In the tail recursive version of the factorial function, we use an accumulator to keep track of the product of the 
numbers as we recursively call the function. This way, we avoid the need for the function to wait for the result of 
the recursive call before performing the multiplication, which is what happens in the head recursive version.
'''

print(fact_tail(5))
````

````python
# Sum of digits of a number using head and tail recursion
def sum_of_digits_head(n):
    if n == 0:
        return 0
    
    small_answer = sum_of_digits_head(n//10)
    final_answer = small_answer + n%10

    return final_answer

def sum_of_digits_tail(n, accumulator=0):
    if n == 0:
        return accumulator
    
    return sum_of_digits_tail(n//10, accumulator + n%10)

print(f"Sum of digits (head recursion): {sum_of_digits_head(12345)}")
print(f"Sum of digits (tail recursion): {sum_of_digits_tail(12345)}")
````

````python
# Power of a number using head and tail recursion
def power_head(base, exp):
    if exp == 0:
        return 1
    
    small_answer = power_head(base, exp-1)
    final_answer = base * small_answer

    return final_answer

def power_tail(base, exp, accumulator=1):
    if exp == 0:
        return accumulator
    
    return power_tail(base, exp-1, accumulator * base)

print(f"Power (head recursion): {power_head(2, 5)}")
print(f"Power (tail recursion): {power_tail(2, 5)}")
````
