#  Recursion

##  Definition

**Recursion** is a programming technique where a function **calls itself** to solve a problem by breaking it into smaller subproblems.

- In simple words:
A function repeats itself until a stopping condition is reached.

---

#  Basic Structure

```python id="g9m2x1"
def function_name():
    if base_condition:
        return value
    else:
        return function_name(smaller_problem)
```

---

#  Important Parts of Recursion

##  1. Base Case

* Condition that stops recursion

##  2. Recursive Case

* Function calls itself with smaller input

---

#  Example: Factorial

```python id="p4v7k3"
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

### Output:

```id="r8y2m6"
120
```

---

#  Example: Fibonacci Series

```python id="t6n1q9"
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))
```

### Output:

```id="u3p8c4"
8
```

---

#  Example: Sum of Natural Numbers

```python id="w5k9d2"
def summation(n):
    if n == 1:
        return 1
    return n + summation(n - 1)

print(summation(5))
```

### Output:

```id="x2m6v7"
15
```

#  How Recursion Works

Example:

```python id="a1q4r8"
factorial(4)
```

Steps:

```id="e7k3p1"
4 * factorial(3)
4 * 3 * factorial(2)
4 * 3 * 2 * factorial(1)
4 * 3 * 2 * 1
= 24
```

---

#  Recursion vs Loop

| Feature                         | Recursion | Loop        |
| ------------------------------- | --------- | ----------- |
| Uses function calls             | Yes       | No          |
| Easier for tree/divide problems | Yes       | No          |
| Faster                          | No        | Usually Yes |
| Memory usage                    | More      | Less        |

---

#  Advantages

1. Clean and short code
2. Good for tree/graph problems
3. Useful in divide-and-conquer algorithms

---

#  Disadvantages

1. More memory usage
2. Slower than loops
3. Can cause stack overflow if no base case

---

#  Recursion Limit

```python id="n8t5x2"
import sys
print(sys.getrecursionlimit())
```

---

#  Important Notes 

* Must have base case
* Recursive call reduces problem size
* Uses call stack memory
* Useful for factorial, Fibonacci, tree traversal

---

#  Summary

* Recursion = function calling itself
* Needs base case + recursive case
* Elegant for many problems
* Slower but useful in complex tasks


````python
# factorial number
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
    
print(factorial(5))
````
